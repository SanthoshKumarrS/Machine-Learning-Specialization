# <span style="font-family:cursive;text-align:center">Reinforcement Learning : </span>

Reinforcement Learning allows an agent to learn how to behave in an environment by trial and error. The agent is rewarded for taking actions that lead to desired outcomes and penalized for taking actions that lead to undesired outcomes. Over time, the agent learns to take the actions that maximize its rewards.

`Appilcations :`

- Controlling Robots
- Factory Optimization
- Financial (stock) Trading
- Playing games (including video games)

`Position of Agent` = State ( s )
`Taking action` = Action ( a )

- `Reward` is given for the desired State and the agent after getting Reward goes into `terminal state` (i.e end of action)

- gamma Discount factor `y` given to the reward,That makes algorithm impatient and get to reward quicker,Without wasting time steps

- return = R1 + Y.R2 = Y^2.R3 + .... it's the sum of discounted rewrds

- Policy ( $Π$ ) takes some inputs ib the present state (s) tells what action to take

a = $Π$(s)

## <span style="font-family:cursive;text-align:center">State-action value function definition : </span>

`State of Value Function (Q Function)` is a function ,denoted as 
`Q(s,a) = Return ` if you

- Start in state s.
- take action a (once)
- then behave optimally after that

## <span style="font-family:cursive;text-align:center">Bellman Equation : </span>

- `s` = current state
- `R(s)` = reward of current state
- `s'` = state you get to after taking action a
- `a'` = action that you take in state s'

**Equation = Q(s,a) = R(s)+Y(maxQ)a'(s',a')**


## <span style="font-family:cursive;text-align:center">Continuous state spaces : </span>

ALgorithm only stores the last 10,000 tuples (steps) it's called `Replay Buffer`

**How to choose actions while still learning ?**

In some state s 
`Option 1 :
- Pick the action `a` that maximizes Q(s,a).
`Option 2:`
- With probability 0.95,pick the action `a` that maximizes Q(s,a).Greedy,"Exploitation"
- With probability 0.05,pick an action `a` randomly. "Exploration"
- Usually increase greedy policy in the tart and gradually decrease it.
