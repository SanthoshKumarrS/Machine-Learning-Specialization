## <span style="font-family:cursive;text-align:center">Clustering  : </span>

In Clustering the algorithms looks at data points and utomatically finds/groups data points that are related or similar to each other.

`Applications :`

* Grouping Similar News
* Market Segmentation
* Analyse DNA data
* Astronomical data Analysis

### <span style="font-family:cursive;text-align:center">K-Means Clustering Algorithm  : </span>

- Randomly intialize K cluster centroids
- Step 1 : Assign Each point to its closest Centroid
- Step 2 : Recompute Centroids (repeat)

### <span style="font-family:cursive;text-align:center">Optimization objective : </span>

- Moto is to minimize the cost function
- After every iteration/step the cost function goes down 
- And if cost function stops minimizing ,the algorithm hhave converged

### <span style="font-family:cursive;text-align:center">Random Initialization : </span>

* Choose K lesser than m (training examples)
* Randomly pick `K` training examples
* Compute cost function (distortion)

### <span style="font-family:cursive;text-align:center">Choosing No of Clusters : </span>

Right no of `K` is often ambiguous
- Run K means with different values of `K` and calculate cost function
- Plot with `Elbow Method` to choose the optimal value for `K`
- Choose `K` based on the use case scenario

## <span style="font-family:cursive;text-align:center">Anamoly Detection : </span>

Anomaly detection algorithms look at an unlabeled dataset of normal events and thereby learns to detect or to raise a red flag for if there is an unusual or an anomalous event

## <span style="font-family:cursive;text-align:center">Anamoly Detection vs Supervised Learning: </span>

`When to Use Anamoly Detection :`

- Very Small number of Positive Examples & Large no of Negative Examples
- Even though we can use supervised learning for low positive examples, Future anamolies may look different from present anamoly,Hard for supervised algorithm to predict it as it looks for positive examples similar to training set
- Anomaly detection finds even different anamolies
