## <span style="font-family:cursive;text-align:center">Making Recommendations & Collaborative Filtering : </span>

There will be users and items to recommend ,Collaborative filtering is a technique used in recommender systems to suggest items to users based on the preferences of other users with similar tastes


## <span style="font-family:cursive;text-align:center">Limitations of Collaborative Filtering : </span>

Cold Start Problem of How to

- rank new items that users have rated?
- show something reasonable to new users who have rated few items?

Use side information about items or users :

- Item : Genre,movie stars,studio...
- User : Demographics (age,gender,location),expressed,preferences,...

## <span style="font-family:cursive;text-align:center">Principle Component Analysis (PCA) : </span>

- It's an algorithm that takes a data with lot features and reduce the size of features in data,Mostly used to visualize the Data.

`Applications of PCA`

* Frequently Used for Visualisation
* Previously used for Data compression
* Speeding up training of a supervised learning Model with algos like SVM
