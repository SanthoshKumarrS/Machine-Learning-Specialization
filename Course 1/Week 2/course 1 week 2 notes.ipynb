## <span style="font-family:cursive;text-align:center">Multiple Linear Regression </span>

`Multiple Features`

Formula for Linear Regression with Multiple Features :  $$w->.x-> +b$$


`Vectorization`

* Using Vectorization makes it easier to write code as well as runs the code faster better than a code in for-loop.
* In a for loop calculations are done one by one whereas in a Vectorized code all the calculations are done in one step.
* Vectorized code runs efficiently and can scale to large datasets.
* Every calculations are done parallely.

`Gradient Descent for Multiple Linear Regression`

**Feature Scaling**

* Feature scaling helps in the niform distribution of data
* should aim for about -1<= & <=1 for each feature
* Finding the weights and params would be easier
* The scaled features will range from 0 to 1
* Mean Normalization, Z-score normalization can also be used for scaling

`Checking gradient descent for convergence`

* J(w,b) should decrease after every iteration 
* It's difficult to say how many iterations are needed for Gradient descent to converge
* Automatic Convergence test ,Let E "epsilon" be $10^-3$ 
* If cost function decreased by <= epsilon in one iteration,declare **Convergence** (as we found params close to global minimum)

`Choosing the learning rate`

* Algorithm runs better with learning rate ($\alpha$) ,too small = slow and too large will fail to converge
* With a small enough Learning rate ,Cost function should decrease on every iteration ,but may take lot of iterations to converge
* `hack` Try with options like **0.001,0.003,0.01,0.03,0.1,0.3,1**
* As every learning rate is only 3 times bigger

`Feature Engineering`

* Features can have huge impact on how Algorithm works
* Adding,removing and creating new features from the dataset is Feature Engineering
* New features can be created by adding or transforming one or more existing features
