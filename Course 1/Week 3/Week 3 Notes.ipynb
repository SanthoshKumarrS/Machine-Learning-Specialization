## <span style="font-family:cursive;text-align:center">Classification with Logistic Regression :</span>

* In classification `Output` can be only one of a handful options
* **y** can be one of two values in `Binary classification`
        ✔ false = 0 / negative / absence
        ✔ true = 1 / positive/presence
* Linear Regression always fits a best line inbetween the target values,incase of outliers the algorithm will misclassify entirely 

`Logistic Regression`

* It's the widely used classification algorithm in the world.
* Logistic Regression fits a `S` shaped curve to the data known as sigmoid function.
* Sigmoid function outputs value between 0 and 1.
* The formula for a sigmoid function is as follows -  

$g(z) = \frac{1}{1+e^{-z}}\$
* Algorithm predicts the percentage of the positive class in percentage.

Alt text

`Decision Boundary`

* The threshold above which predicted values = 1 and below it is = 0 is Decision boundary,
* It's is a line that divides the feature space into two regions: one region where the model predicts class 0 and another region where it predicts class 1. The decision boundary is determined by the logistic regression model's coefficients (weights) and the bias term.
* Decision boundary cannot always be non-linear with data containg complex features.

## <span style="font-family:cursive;text-align:center">Cost function for logistic regression :</span>

`Cost Function` 

* Squared error cost function isn't idle for logistic regression.
* Cost function is non-convex & there are lots of local minimum & can be difficult.
* when y =1 & algorithm predicts close to 1 loss is low and if algorithm predicts closer to 0 loss is infinite.
* when y =0 & algorithm predicts close to 0 loss is low and if algorithm predicts closer to 1 loss is infinite.
* The further preiction is from the target (y),the higher the loss.

`Simplified Cost Function`

$$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \$$

## <span style="font-family:cursive;text-align:center">Gradient Descent for logistic regression :</span>

`Gradient Descent Implementation :`

$$\begin{align*}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \ \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \ 
\end{align*}$$

        ☠ The update steps look like the update steps for linear regression, but the definition of Model is different.


* m is the number of training examples in the data set      
* $f_{\mathbf{w},b}(x^{(i)})$ is the model's prediction, while $y^{(i)}$ is the target
* For a logistic regression model  
    $z = \mathbf{w} \cdot \mathbf{x} + b$  
    $f_{\mathbf{w},b}(x) = g(z)$  
    where $g(z)$ is the sigmoid function:  
    $g(z) = \frac{1}{1+e^{-z}}$   
    


## <span style="font-family:cursive;text-align:center">The Problem of Overfitting :</span>

`Overfittng`

* Fits the training data Extremely well,will work on training data,But fails on new / unseen data
* cost function will be zero or close to zero
* Algorithm will have high variance.


`Underfitting`

* The Algorithm does not fit the training data well and have high bias. 

`Right Model`
* Just right model fits pretty well on the training data and is genaralized.


This can be compared with a soup bowl `underfit model` is too cold while `overfit model` is too hot and `Generalized model` is right to consume 

`Addressing overfitting`

* Collect more Training Data.
* Reducing the features in the dataset by Using Feature Selection
* Because lot of features combined with small no of data often results in Overfitting
* Regularization can also help to reduce overfitting ( reduce the size of parameters)

`Cost function with regularization`

Regularization is a technique used to prevent overfitting by adding extra information to the dataset or by shrinking the coefficients towards zero. This discourages the learning of a more complex or more flexible machine learning model

$$\frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2 \$$ 

Lambda balances the fitting the data and keeping the parameters small,if lamda is too small parameters become zero

`Regularized linear regression`

$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2  + \frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2 \$$ 

`Regularized Logistic regression`

$$J(\mathbf{w},b) = \frac{1}{m}  \sum_{i=0}^{m-1} \left[ -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \right] + \frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2 \$$

The gradient calculation for both linear and logistic regression are nearly identical, differing only in computation of $f_{\mathbf{w}b}$.
$$\begin{align*}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)}  +  \frac{\lambda}{m} w_j \ \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \ 
\end{align*}$$
