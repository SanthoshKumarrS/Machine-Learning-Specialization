### <span style="font-family:cursive;text-align:center"> Train the model with gradient descent : </span>

`Gradient Descent`

gradient descent formula 

* Gradient descent is an algorithm for finding values of parameters w and b that minimize the cost function J
* Here `w` and `b` are updated simultaneously
* Convex function is a bowl shaped cost function ,that have only one global minimum and no local minimum

In lecture, *gradient descent* was described as:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \  \; \newline 
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$
where, parameters $w$, $b$ are updated simultaneously.  
The gradient is defined as:
$$
\begin{align}
\frac{\partial J(w,b)}{\partial w}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)}) \\\
\end{align}
$$

Here *simultaniously* means that you calculate the partial derivatives for all the parameters before updating any of the parameters.

`Learning Rate`

* Learning rate is something that decides how big the step is taken by the algorithm
* Learning rate is always positive
* If it's too small the algorithm takes very small step ,eventhough it reduces the cost function ,the algorithm will be slow and will take very long time
        ✔ Algorithm may be too small
* If it's too large the algorithm takes big strides and the algorithm may miss the Minimum and will increase the cost function
        ✔ Overshoot, never reach minimum
        ✔ Fail to  converge and may diverge
