# <span style="font-family:cursive;text-align:center">Decision Trees 🌳 : </span>

`Decision Tree`  is a tree-like structure where each internal node denotes a test on an attribute, each branch represents an outcome of the test, and each leaf node holds a class label or a prediction value.

## <span style="font-family:cursive;text-align:center">Learning Process : </span>

- Need to decide what feature is in the root node
- When do you stop splitting
        * When node is 100% one class
        * When splitting a node will result in tree exceeding a maximum depth
        * when improvements in purity score are below a threshold
        * when number of examples in a node is below a threshold

## <span style="font-family:cursive;text-align:center">Decision tree learning : </span>

### <span style="font-family:cursive;text-align:center">Measuring Purity : </span>


`Entropy` is used as a measure of impurity or disorder within a dataset. It plays a crucial role in deciding how to split data at each node of the decision tree

### <span style="font-family:cursive;text-align:center">Choosing a split: Information Gain : </span>

- Reduction of Entrophy is called `Information Gain`
- Which feature have the best reduction in entrophy score becomes the root node and so on

- Start with all examples at the root node
- Calculate information gain for all possible features & pick the one with the highest information gain
- spilt dataset according to selected feature and create left and right branch of the tree
- keep repareating until stopping criteria is met

### <span style="font-family:cursive;text-align:center">Using one-hot encoding of categorical features
 : </span>
When there is more than 2 categorical features ,We can use one-hot encoding

## <span style="font-family:cursive;text-align:center">Tree Ensembles : </span>

### <span style="font-family:cursive;text-align:center">Using multiple decision trees : </span>

We get better results when we train a bunch of trees,rather than a Single Tree

### <span style="font-family:cursive;text-align:center">Random Forest & XGBoost : </span>

Given training set of size `m`, For b = 1 to B:

- Use Sampling with replacement to create a new training set of size `m`,Train a decision tree on the new Dataset (random forest)

- Instead of picking from all exampleswith equal probability,make it more likely to pick misclassified examples from previously Trained Trees.
- Xgboost is opensource implementation of boosted trees
- Built in regularization to prevent overfitting
- Fast & Efficient implementation

### <span style="font-family:cursive;text-align:center">When to use Decision trees : </span>

`Decision Trees & Tree Ensemnles`
* Works well on tabular(structured data)
* Not recommended on Unstructured data (images,audio,text)
* Faster to train
* Small decision trees may be human interpretable

`Neural Networks`
* Works well on all types of data including tabular and unstructured data
* May be slower than decision tree
* Works with transfer learning
* When building a system of multiple models working together ,it might be easier to string together multiple neural networks.
