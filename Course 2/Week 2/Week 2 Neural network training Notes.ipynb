## <span style="font-family:cursive;text-align:center">Activation Functions </span>

In Neural Networks, an activation function is a function that is applied to the output of a neuron to determine whether the neuron should be activated or not. 

`Alternatives to the sigmoid activation`

* **Relu** is the most commonly used Activation Function (**Re**ctified **L**inear **U**nit)
* Linear Activation function ( `g(Z) = Z` ) is like not using an activation function
* Sigmoid,Relu,Linear & Softmax are most commonly used activation functions

`Choosing an Activation Functions`

* For a Binaryclassification Sigmoid is Mostly used at output layer
* For regression Linear activation & for non negative linear function Relu can be used at output layer
* For Hidden layers Relu is the most commonly used one
    ✔ Relu is Faster to complete compared to sigmoid & Faster learning
    ✔ sigmoid flattens in two parts compared to Relu where its only one and Gradient descent will be slow.
    
`Why do we Need Activation Functions`

Activation functions are used to introduce non-linearity into neural networks, which allows them to learn more complex relationships between input and output data.

Without Activation Functions ,neural netwroks performs like ML(linear/logistic Regression) Algorithms

## <span style="font-family:cursive;text-align:center">Multiclass Classification </span>

`Multiclass`

It refers to the problem where output can be more than 2 possible outcomes

`Neural Network with Softmax`

* Output neuron should be no of Outputs expected 
* Activation function is **Softmax**
* loss function is **SparseCategoricalCrossentropy**
* from_logits = True argument helps to reduce the numerical round off error 
* this means the output layer becomes **linear** activation function

## <span style="font-family:cursive;text-align:center">Additional Neural Network Concepts </span>

`Advanced Optimization`

* **ADAM** stands for Adaptive Moment Estimation, and it is a method for calculating adaptive learning rates for stochastic optimization problems.
* It uses different learning rate for every single parameter of the model.
* It needs some default initial learning rate and it can be set as le-3 ,but can be experimented

`Additional Layer Types`

* For some applications ,few neural networks use a different layer than a Dense one ,for eg : `Convolutional Layer`
* In Convolutional Layer every neuron looks only at certain part of the previous layer's inputs 
        ✔ It's Faster to compute,
        ✔ Need less training data,
        ✔ less prone to overfitting
* If there are lot of Convolutional Layer then its a Convolutional Neural Network 

## <span style="font-family:cursive;text-align:center">Back Propagation</span>

`Derivative`

TensorFlow will automatically use back propagation to compute derivatives and use gradient descent or Adam to train the parameters of a neural network

Derivate is just if w goes up by a little bit, by Epsilon, how much does J of w change as a consequence

* Backprop is an efficient way to calculate derivatives
