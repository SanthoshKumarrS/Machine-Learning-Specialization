### <span style="font-family:cursive;text-align:center">Model selection and training/cross validation/test sets : </span>


Cross-validation is a statistical method used to estimate the performance of a machine learning model on unseen data. It is used to protect against overfitting, which is when a model learns the training data too well and is unable to generalize to new data.

Cross-validation works by dividing the training data into multiple folds. Each fold is then used as a test set, and the model is trained on the remaining folds. The model's performance on the test set is then recorded. This process is repeated for each fold, and the average performance across all folds is used as an estimate of the model's overall performance.

* Cross-validation also known as dev (Development) set.
* Pick the model whic has lowest cross-validation loss.
* Finalize the model and then come and test it with the test set
* Bcoz that ensures that your test set is a fair and not overly optimistic estimate of how well your model will generalize to new data

`split your data into three parts:`

* training set - used to train the model
* cross validation set (also called validation, development, or dev set) - used to evaluate the different model configurations you are choosing from. For example, you can use this to make a decision on what polynomial features to add to your dataset.
* test set - used to give a fair estimate of your chosen model's performance against new examples. This should not be used to make decisions while you are still developing the models.

## <span style="font-family:cursive;text-align:center">Bias and variance : </span>

`High Bias` = Underfitting & `High Variance` = Overfitting

* In High Bias - loss in training & dev set is high
* In High Variance - loss in training is low & dev set is high (performs poorly on unseen data)

`High Bias & High Variance` is loss on traing set will be high and Validatation set loss will be much higher


### <span style="font-family:cursive;text-align:center">Regularization on Bias and variance : </span>

* Choise of Regularization parameter affects bias & variance
* Lower `r` high variance and higher `r`high bias

### <span style="font-family:cursive;text-align:center">Learning Curve : </span>

Learning curves are a way to help understand how your learning algorithm is doing as a function of the amount of experience it has

`Debugging a Learning Algorithm`

To fix a high bias problem, you can:

* try adding polynomial features
* try getting additional features
* try decreasing the regularization parameter

To fix a high variance problem, you can:

* try increasing the regularization parameter
* try smaller sets of features
* get more training examples

## <span style="font-family:cursive;text-align:center">ML Development Process : </span>

### <span style="font-family:cursive;text-align:center">Error Analysis : </span>

- Manually Examining few examples of the missclassified by algorithmss in order to identify common traits and trends and checking the impact each type of error causes
- carrying out this form of error analysis to be really helpful to screening or to deciding which changes to the model are more promising to try on next

`Limitations :`
Difficult to decide for Tasks that human's can't do eg: Ads Recomendation

### <span style="font-family:cursive;text-align:center">Adding Data : </span>

`Data Augmentation :` Modifying an existing training example to create a new training example,Augumented Examples should be representative or quite similar of the unseen data in test data,No use in adding unwanted data,Mostly used for Computer Vision Tasks

### <span style="font-family:cursive;text-align:center"> Transfer Learning: </span>

- Option 1 - Only train output layers parameters (supervised pretraining), download neural network with parameters that have been pre-trained on a large dataset with the same input type as your application
- Option 2 - Train all parameters.(fine tuning),Further train the network on your own data.

`Limitations :` Input type should be same (Dimensions and types like text,Audio,Image)
